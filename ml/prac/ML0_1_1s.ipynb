# генераторные выражения с условиями и циклами

В Python, когда вы создаёте объекты, такие как списки, словари, или даже числа, они занимают определённое количество памяти. Чем больше данных хранится в объекте, тем больше памяти он занимает. Если вы создаете список с миллионом элементов, все эти элементы будут храниться в памяти одновременно.

```python
squares_list = [x**2 for x in range(1000000)]
```

```python
squares_list
```

```python
squares_gen = (x**2 for x in range(1000000))
```

```python
squares_gen
```

```python
print(next(squares_gen))
```

```python
for square in squares_gen:
  print(square)
```

Генерация квадратов чётных чисел в диапазоне

```python
even_squares = (x**2 for x in range(10) if x % 2 == 0)

for square in even_squares:
    print(square)
```

Генерация произведений чисел из двух списков

```python
list1 = [1, 2, 3]
list2 = [4, 5, 6]
products = (x * y for x in list1 for y in list2)

for product in products:
    print(product)
```

# анонимные функции

Анонимные функции или lambda — это небольшие функции, которые создаются с использованием ключевого слова lambda и не требуют определения имени.

Использование lambda для сортировки списка по длине строк

```python
words = ['apple', 'banana', 'cherry', 'date']

sorted_words = sorted(words, key=lambda word: len(word))

print(sorted_words)
```

# функция генерации сетки mgrid

```python
a, b = np.mgrid[1:5, 0:5]
print(a)
print(b)
print(a.shape)
print(b.shape)
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

x, y = np.mgrid[-5:5:100j, -5:5:100j]  # Создание сетки от -5 до 5 со 100 точками
print(x)
print(y)
print(x.shape)
print(y.shape)
z = (x**2 + y**2)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(x, y, z, cmap='viridis')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
plt.show()
```

# поэлементные операции

```python
list1 = [1, 2, 3]
list2 = [4, 5, 6]

sum_list = [x + y for x, y in zip(list1, list2)]
print(sum_list)
```

```python
list_numbers = [1, 2, 3, 4, 5]

doubled_evens = [x * 2 for x in list_numbers if x % 2 == 0]
print(doubled_evens)
```
