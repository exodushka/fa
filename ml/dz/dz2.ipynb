```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('data.csv')
```

```python
rooms_counts = df['Rooms'].value_counts().sort_index()
explode = [0.1 if i == rooms_counts.idxmax() else 0 for i in rooms_counts.index]
plt.figure(figsize=(9, 7))

plt.pie(rooms_counts,
        labels=[f'{int(r)} комн.' for r in rooms_counts.index],
        explode=explode,
        textprops={'fontsize': 11, 'weight': 'bold'})

plt.title('Распределение квартир по количеству комнат')
plt.legend(title='Комнат', loc='center left', bbox_to_anchor=(1, 0.5))
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Price'])

plt.title('Гистограмма распределения цены квартир', fontsize=14, weight='bold')
plt.xlabel('Цена, руб.', fontsize=12)
plt.ylabel('Количество квартир', fontsize=12)

plt.show()
```

```python
plt.scatter(df["Rooms"], df["Price"])

plt.title('Price от Rooms')
plt.xlabel('Кол-во комнат')
plt.ylabel('Цена')
k, b = np.polyfit(df["Rooms"], df["Price"], deg=1)
trend_line = k * df["Rooms"] + b
plt.plot(df["Rooms"], trend_line, color="red")

plt.show()
```

```python
plt.scatter(df["Square"], df["Price"])

plt.title('Price от Square')
plt.xlabel('Площадь')
plt.ylabel('Цена')
k, b = np.polyfit(df["Square"], df["Price"], deg=1)
trend_line = k * df["Square"] + b
plt.plot(df["Square"], trend_line, color="red")

plt.show()
```

```python
plt.scatter(df["HouseFloor"], df["Price"])

plt.title('Price от HouseFloor')
plt.xlabel('Кол-во этажей')
plt.ylabel('Цена')
k, b = np.polyfit(df["HouseFloor"], df["Price"], deg=1)
trend_line = k * df["HouseFloor"] + b
plt.plot(df["HouseFloor"], trend_line, color="red")

plt.show()
```

```python
plt.scatter(df["HouseYear"], df["Price"])

plt.title('Price от HouseYear')
plt.xlabel('Год')
plt.ylabel('Цена')
plt.xlim(1800, 2200)
k, b = np.polyfit(df["HouseYear"], df["Price"], deg=1)
trend_line = k * df["HouseYear"] + b
plt.plot(df["HouseYear"], trend_line, color="red")

plt.show()
```

```python
plt.figure()
sns.kdeplot(df['Price'])
plt.title('Ядерная оценка плотности цены')
plt.xlabel('Цена')
plt.ylabel('Плотность')
plt.show()
```

```python
plt.figure()
sns.kdeplot(data=df, x='HouseFloor', y='Price', fill=True)

plt.title('Двумерная ядернуая оценка плотности Price и HouseFloor', fontsize=14, weight='bold')
plt.xlabel('Этаж')
plt.ylabel('Цена')
plt.tight_layout()
plt.show()
```

```python
plt.figure()
sns.boxplot(x=df['Square'])
plt.title('Ящиковая диаграмма Square')
plt.xlabel('Площадь')
plt.show()
```
